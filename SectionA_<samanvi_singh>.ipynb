```python
# question 33
is_prime = True
if n <= 1:
    is_prime = False
else:
    i = 2
    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break
        i = i + 1

if is_prime:
    print(n, "is prime")
else:
    print(n, "is not prime")

prime_sum = 0
for num in range(2, n + 1):
    check_prime = True
    j = 2
    while j * j <= num:
        if num % j == 0:
            check_prime = False
            break
        j = j + 1

    if check_prime:
        prime_sum = prime_sum + num

print("Sum of primes:", prime_sum)
```

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        total_amount = float(amount)
        base_amount = total_amount / (1 + (gst_rate / 100))
        gst_amount = total_amount - base_amount
    else:
        base_amount = float(amount)
        gst_amount = base_amount * (gst_rate / 100)
        total_amount = base_amount + gst_amount

    return (round(base_amount, 2), round(gst_amount, 2), round(total_amount, 2))

print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

```python
# question 47
def power(base, exp):
    if exp == 0:
        return 1
    elif exp < 0:
        return 1 / power(base, -exp)
    else:
        return base * power(base, exp - 1)

print(power(2, 10))
print(power(5, 0))
print(power(3, -2))
```

```python
# question 48
import numpy as np

matrix = np.eye(5)
print("Original 5x5 Identity Matrix:")
print(matrix)

matrix_10 = matrix * 10
print("\nMultiplied by 10:")
print(matrix_10)

matrix_t = np.transpose(matrix_10)
print("\nTransposed Matrix:")
print(matrix_t)

diag_sum = np.trace(matrix_t)
print("\nSum of diagonal elements:", diag_sum)

total_sum = np.sum(matrix_t)
print("Sum of all elements:", total_sum)
```

```python
# question 49
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"])

total_sales = np.sum(sales)
avg_sales = np.mean(sales)
max_day = days[np.argmax(sales)]
above_avg = np.sum(sales > avg_sales)
growth = ((sales[-1] - sales[0]) / sales[0]) * 100
dod_changes = (np.diff(sales) / sales[:-1]) * 100

print("Total weekly sales:", total_sales)
print("Daily average:", round(avg_sales, 2))
print("Maximum sales day:", max_day)
print("Days above weekly average:", above_avg)
print("Percentage growth Mon to Sun:", round(growth, 2))
print("Day-on-day % changes:", np.round(dod_changes, 2))
```
